```python
# ================================================================
# 2. 📂 ĐỌC DỮ LIỆU
# ================================================================

cty_hangtau = pd.read_excel(
    '/content/drive/MyDrive/KT/Data/Hãng tàu.xlsx'
)

chuhang_hangtau = pd.read_excel(
    '/content/drive/MyDrive/KT/Data/Chủ hàng.xlsx'
)

salan = pd.read_excel(
    '/content/drive/MyDrive/KT/Data/Sà lan.xlsx'
)


# ================================================================
# 3. 🎨 MÀU SẮC
# ================================================================

dm_TABLE_COLORS = {

    'cty_hangtau': '#E33539',

    'chuhang_hangtau': '#EE6457',

    'salan': '#FF9190'
}


# ================================================================
# 4. 📁 ĐƯỜNG DẪN FILE
# ================================================================

dm_TABLE_FILES = {

    'cty_hangtau':
        '/content/drive/MyDrive/KT/Data/Hãng tàu.xlsx',

    'chuhang_hangtau':
        '/content/drive/MyDrive/KT/Data/Chủ hàng.xlsx',

    'salan':
        '/content/drive/MyDrive/KT/Data/Sà lan.xlsx'
}


# ================================================================
# 5. ⚙️ CẤU HÌNH
# ================================================================

dm_TABLE_CONFIG = {

    'cty_hangtau': {

        'caption':
            '🏢 DANH SÁCH TÊN CÔNG TY - HÃNG TÀU',

        'column_widths': {

            'STT': '70px',

            'Tên công ty': '500px',

            'Mã Bravo': '150px',

            'Mã MISA': '150px'
        }
    },

    'chuhang_hangtau': {

        'caption':
            '🚢 DANH SÁCH TÊN CHỦ HÀNG - HÃNG TÀU',

        'column_widths': {

            'STT': '70px',

            'Chủ hàng': '300px',

            'Hãng tàu': '300px'
        }
    },

    'salan': {

        'caption':
            '🚤 DANH SÁCH SÀ LAN',

        'column_widths': {

            'STT': '70px',

            'Sà lan': '200px',

            'Viết tắt': '150px',

            'MISA': '150px'
        }
    }
}


# ================================================================
# 6. 🔢 HÀM RESET STT
# ================================================================

def dm_reset_stt(df):

    df = df.copy()

    if 'STT' in df.columns:

        df['STT'] = range(
            1,
            len(df) + 1
        )

    return df


# ================================================================
# 7. 🔄 TRẠNG THÁI
# ================================================================

dm_TABLE_DATA = {

    'cty_hangtau':
        dm_reset_stt(cty_hangtau.copy()),

    'chuhang_hangtau':
        dm_reset_stt(chuhang_hangtau.copy()),

    'salan':
        dm_reset_stt(salan.copy())
}


dm_SEARCH_RESULTS = {

    key:
        df.copy()

    for key, df in dm_TABLE_DATA.items()
}


dm_SELECTED_ROWS = {

    'cty_hangtau': None,

    'chuhang_hangtau': None,

    'salan': None
}


dm_UI_CACHE = {}


# ================================================================
# 8. 🔐 BẢO VỆ CẤU TRÚC CỘT GỐC
# ================================================================

dm_TABLE_COLUMNS = {

    key:
        list(df.columns)

    for key, df in dm_TABLE_DATA.items()
}


def dm_preserve_table_columns(key, df):

    df = df.copy()

    original_columns = dm_TABLE_COLUMNS[key]

    for column in original_columns:

        if column not in df.columns:

            df[column] = pd.NA

    df = df[original_columns]

    return df


for key in dm_TABLE_DATA:

    dm_TABLE_DATA[key] = dm_preserve_table_columns(
        key,
        dm_TABLE_DATA[key]
    )

    dm_TABLE_DATA[key] = dm_reset_stt(
        dm_TABLE_DATA[key]
    )

    dm_SEARCH_RESULTS[key] = (
        dm_TABLE_DATA[key].copy()
    )


# ================================================================
# 9. 🔎 SEARCH INDEX
# ================================================================

dm_SEARCH_INDEX = {}


for key, df in dm_TABLE_DATA.items():

    dm_SEARCH_INDEX[key] = (

        df
        .fillna('')
        .astype(str)
        .apply(
            lambda col:
                col.str.lower()
        )
    )


# ================================================================
# 10. 🎨 MÀU BẢNG
# ================================================================

def dm_get_table_color(key):

    return dm_TABLE_COLORS.get(
        key,
        '#E33539'
    )


# ================================================================
# 11. 🔄 REFRESH SEARCH INDEX
# ================================================================

def dm_refresh_search_index(key):

    global dm_SEARCH_INDEX

    dm_SEARCH_INDEX[key] = (

        dm_TABLE_DATA[key]
        .fillna('')
        .astype(str)
        .apply(
            lambda col:
                col.str.lower()
        )
    )


# ================================================================
# 12. 🎨 CSS DANH MỤC
# ================================================================

dm_CSS = """

<style>

/* ============================================================
   🌐 NỀN
   ============================================================ */

html,
body,
.jp-Notebook,
.jp-NotebookPanel,
#notebook-container,
.jp-Cell,
.jp-Cell-outputWrapper,
.jp-OutputArea {

    background-color: #FFE5D8 !important;
}


/* ============================================================
   📦 CONTAINER
   ============================================================ */

.management-container {

    width: 95% !important;

    max-width: 1500px !important;

    margin: 0 auto !important;

    padding: 3px 10px 10px 10px !important;

    background-color: #FFE5D8 !important;

    border-radius: 16px !important;

    box-sizing: border-box !important;

    overflow: visible !important;

    height: auto !important;

    min-height: 0 !important;

    max-height: none !important;
}


/* ============================================================
   📌 TIÊU ĐỀ
   ============================================================ */

.category-title {

    width: 100% !important;

    box-sizing: border-box !important;

    background-color: #000000 !important;

    color: #FFE5D8 !important;

    font-size: 26px !important;

    font-weight: bold !important;

    text-align: center !important;

    padding: 11px !important;

    margin: 3px auto 8px auto !important;

    border-radius: 14px !important;

    box-shadow:
        0 3px 8px rgba(0,0,0,0.18) !important;
}


/* ============================================================
   🔘 3 BUTTON DANH MỤC
   ============================================================ */

.main-menu-container {

    width: 100% !important;

    background-color: #FFE5D8 !important;

    padding: 4px 10px 5px 10px !important;

    margin: 0 auto 5px auto !important;

    border-radius: 16px !important;

    box-sizing: border-box !important;

    gap: 14px !important;

    overflow: visible !important;
}


.main-menu-button {

    width: 200px !important;

    min-width: 200px !important;

    max-width: 200px !important;

    height: 60px !important;

    min-height: 60px !important;

    max-height: 60px !important;

    flex: 0 0 200px !important;

    font-size: 20px !important;

    font-weight: bold !important;

    border-radius: 14px !important;

    border: 2px solid rgba(0,0,0,0.15) !important;

    box-shadow:
        0 4px 8px rgba(0,0,0,0.18),
        inset 0 1px 0 rgba(255,255,255,0.25) !important;

    cursor: pointer !important;

    flex-shrink: 0 !important;

    transition:
        transform 0.2s ease,
        filter 0.2s ease,
        box-shadow 0.2s ease !important;
}


.main-menu-button:hover {

    transform: translateY(-3px) !important;

    filter: brightness(1.05) !important;

    box-shadow:
        0 6px 12px rgba(0,0,0,0.22),
        inset 0 1px 0 rgba(255,255,255,0.25) !important;
}


/* ============================================================
   🛠️ TẤT CẢ BUTTON CHỨC NĂNG
   ⭐ TỰ ĐỘNG THEO ĐỘ DÀI NỘI DUNG
   ============================================================ */

.management-toolbar button,
.management-form button,
.management-message button,
.dm-table-navigation button {

    width: auto !important;

    min-width: 0 !important;

    max-width: none !important;

    height: 38px !important;

    min-height: 38px !important;

    max-height: 38px !important;

    flex:
        0 0 auto !important;

    flex-shrink: 0 !important;

    padding:
        0 18px !important;

    margin: 0 !important;

    background-color: #FFFFFF !important;

    background: #FFFFFF !important;

    color: #000000 !important;

    border: 1.5px solid #000000 !important;

    border-radius: 14px !important;

    box-shadow:
        0 4px 8px rgba(0,0,0,0.18),
        inset 0 1px 0 rgba(255,255,255,0.25) !important;

    cursor: pointer !important;

    box-sizing: border-box !important;

    transition:
        transform 0.2s ease,
        filter 0.2s ease,
        box-shadow 0.2s ease !important;
}


/* ============================================================
   📝 LABEL TRONG BUTTON
   ⭐ KHÔNG CHO TEXT BỊ ÉP RỘNG
   ============================================================ */

.management-toolbar button .widget-label,
.management-form button .widget-label,
.management-message button .widget-label,
.dm-table-navigation button .widget-label {

    width: auto !important;

    min-width: 0 !important;

    max-width: none !important;

    white-space: nowrap !important;

    overflow: visible !important;

    text-align: center !important;

    display: block !important;
}


/* ============================================================
   🛠️ HOVER BUTTON CHỨC NĂNG
   ============================================================ */

.management-toolbar button:hover,
.management-form button:hover,
.management-message button:hover,
.dm-table-navigation button:hover {

    background-color: #FFFFFF !important;

    background: #FFFFFF !important;

    color: #000000 !important;

    transform: translateY(-3px) !important;

    filter: brightness(1.05) !important;

    box-shadow:
        0 6px 12px rgba(0,0,0,0.22),
        inset 0 1px 0 rgba(255,255,255,0.25) !important;
}


/* ============================================================
   🛠️ TOOLBAR
   ============================================================ */

.management-toolbar {

    width: 100% !important;

    background-color: #FFE5D8 !important;

    padding: 4px 10px !important;

    margin: 2px auto 5px auto !important;

    border-radius: 12px !important;

    box-sizing: border-box !important;

    gap: 8px !important;

    height: auto !important;

    min-height: 42px !important;

    max-height: none !important;

    overflow: visible !important;

    flex: 0 0 auto !important;
}


/* ============================================================
   🔘 BUTTON TOOLBAR
   ============================================================ */

.management-toolbar button {

    font-size: 13px !important;

    font-weight: 600 !important;

    white-space: nowrap !important;
}


/* ============================================================
   🔎 Ô TÌM KIẾM
   ============================================================ */

.management-toolbar input,
.management-toolbar .widget-text input {

    width: 180px !important;

    min-width: 180px !important;

    max-width: 180px !important;

    height: 38px !important;

    min-height: 38px !important;

    max-height: 38px !important;

    box-sizing: border-box !important;

    color: #000000 !important;

    background-color: #FFFFFF !important;

    font-size: 14px !important;

    font-weight: 500 !important;

    border: 1.5px solid #000000 !important;

    border-radius: 14px !important;

    padding: 6px 10px !important;

    outline: none !important;

    box-shadow:
        0 4px 8px rgba(0,0,0,0.18) !important;
}


.management-toolbar input:focus,
.management-toolbar .widget-text input:focus {

    border:
        1.5px solid #000000 !important;

    box-shadow:
        0 0 5px rgba(0,0,0,0.20) !important;
}


/* ============================================================
   💬 MESSAGE
   ============================================================ */

.management-message {

    width: 100% !important;

    background-color: #FFE5D8 !important;

    color: #000000 !important;

    margin: 0 !important;

    padding: 0 !important;

    box-sizing: border-box !important;

    text-align: center !important;

    height: auto !important;

    min-height: 0 !important;

    max-height: none !important;

    overflow: visible !important;

    flex: 0 0 auto !important;

    line-height: 1.2 !important;
}


.management-message > .widget-output,
.management-message .widget-output,
.management-message .output_area,
.management-message .output_subarea,
.management-message .jp-OutputArea,
.management-message .jp-OutputArea-child {

    width: 100% !important;

    height: auto !important;

    min-height: 0 !important;

    max-height: none !important;

    overflow: visible !important;

    background: transparent !important;

    margin: 0 !important;

    padding: 0 !important;

    box-sizing: border-box !important;
}


.management-message pre {

    background: transparent !important;

    color: #000000 !important;

    font-family: Arial, sans-serif !important;

    font-size: 15px !important;

    font-weight: 600 !important;

    text-align: center !important;

    white-space: pre-wrap !important;

    margin: 0 !important;

    padding: 0 !important;

    line-height: 1.2 !important;
}


.management-message p {

    margin: 0 !important;

    padding: 0 !important;

    line-height: 1.2 !important;
}


/* ============================================================
   📝 FORM THÊM / SỬA
   ============================================================ */

.management-form {

    width: 100% !important;

    background-color: #FFE5D8 !important;

    color: #000000 !important;

    border-radius: 12px !important;

    padding: 5px !important;

    margin: 2px auto 5px auto !important;

    box-sizing: border-box !important;

    height: auto !important;

    min-height: 0 !important;

    max-height: none !important;

    overflow: visible !important;

    flex: 0 0 auto !important;

    align-self: stretch !important;
}


.management-form > .widget-vbox {

    width: 100% !important;

    height: auto !important;

    min-height: 0 !important;

    max-height: none !important;

    overflow: visible !important;
}


/* ============================================================
   📝 INPUT FORM
   ============================================================ */

.management-form .widget-text {

    width: 650px !important;

    height: 38px !important;

    min-height: 38px !important;

    max-height: 38px !important;

    margin: 2px auto !important;

    overflow: visible !important;

    flex: 0 0 38px !important;
}


.management-form .widget-text input {

    width: 100% !important;

    height: 38px !important;

    min-height: 38px !important;

    box-sizing: border-box !important;

    background-color: #FFFFFF !important;

    color: #000000 !important;

    border: 1.5px solid #000000 !important;

    border-radius: 14px !important;

    padding: 5px 10px !important;

    outline: none !important;

    box-shadow:
        0 4px 8px rgba(0,0,0,0.12) !important;
}


.management-form .widget-text input:focus {

    border:
        1.5px solid #000000 !important;

    box-shadow:
        0 0 5px rgba(0,0,0,0.15) !important;
}


.management-form .widget-label {

    color: #000000 !important;

    font-size: 14px !important;

    font-weight: 600 !important;
}


/* ============================================================
   🔘 BUTTON FORM
   ============================================================ */

.management-form button {

    font-size: 13px !important;

    font-weight: 600 !important;

    white-space: nowrap !important;
}


.management-form .form-button-box {

    width: 100% !important;

    height: 45px !important;

    min-height: 45px !important;

    margin: 6px 0 3px 0 !important;

    display: flex !important;

    justify-content: center !important;

    align-items: center !important;

    gap: 10px !important;

    overflow: visible !important;

    flex: 0 0 45px !important;
}


/* ============================================================
   📊 BẢNG
   ============================================================ */

.management-table-scroll {

    width: 100% !important;

    height: 500px !important;

    min-height: 500px !important;

    max-height: 500px !important;

    overflow: auto !important;

    background-color: #FFE5D8 !important;

    border-radius: 14px !important;

    margin: 0 auto !important;

    padding: 3px !important;

    box-sizing: border-box !important;

    border: 2px solid rgba(0,0,0,0.12) !important;

    box-shadow:
        0 4px 10px rgba(0,0,0,0.12) !important;

    flex: 0 0 500px !important;
}


.management-table-scroll .output_area,
.management-table-scroll .output_subarea {

    overflow: visible !important;

    max-width: none !important;

    background: transparent !important;

    margin: 0 !important;

    padding: 0 !important;
}


.management-table-scroll table {

    margin: 0 auto !important;

    border-collapse: separate !important;

    border-spacing: 0 !important;

    border-radius: 10px !important;

    overflow: hidden !important;
}


.management-table-scroll table thead th {

    position: sticky !important;

    top: 0 !important;

    z-index: 20 !important;
}


/* ============================================================
   🔙 NAVIGATION
   ============================================================ */

.dm-table-navigation {

    width: 100% !important;

    display: flex !important;

    flex-direction: row !important;

    justify-content: flex-start !important;

    align-items: center !important;

    gap: 8px !important;

    margin: 0 0 6px 0 !important;

    padding: 0 !important;

    box-sizing: border-box !important;

    height: 38px !important;

    min-height: 38px !important;

    overflow: visible !important;

    flex: 0 0 38px !important;
}


.dm-table-navigation button {

    font-size: 13px !important;

    font-weight: 600 !important;

    white-space: nowrap !important;

    display: flex !important;

    align-items: center !important;

    justify-content: center !important;
}


/* ============================================================
   🔙 NAVIGATION CỦA MENU DANH MỤC
   ============================================================ */

.main-header .dm-table-navigation button {

    font-weight: bold !important;
}


/* ============================================================
   📦 VBOX
   ============================================================ */

.management-container .widget-vbox,
.management-container .widget-box {

    overflow: visible !important;

    flex-shrink: 0 !important;

    height: auto !important;

    min-height: 0 !important;

    max-height: none !important;
}


/* ============================================================
   🔥 OUTPUT MÀN HÌNH
   ============================================================ */

.app-view-output {

    width: 100% !important;

    height: auto !important;

    min-height: 0 !important;

    max-height: none !important;

    overflow: visible !important;

    margin: 0 !important;

    padding: 0 !important;
}


.app-view-output > .output_area,
.app-view-output > .output_subarea,
.app-view-output .jp-OutputArea,
.app-view-output .jp-OutputArea-child {

    height: auto !important;

    min-height: 0 !important;

    max-height: none !important;

    overflow: visible !important;

    margin: 0 !important;

    padding: 0 !important;
}


/* ============================================================
   🔘 APP NAV BUTTON
   ============================================================ */

.app-nav-button {

    box-sizing: border-box !important;
}

</style>
"""


# ================================================================
# 13. 🎨 CSS WIDGET
# ================================================================

dm_css_widget = widgets.HTML(

    value=dm_CSS,

    layout=widgets.Layout(

        width='0px',

        height='0px',

        min_width='0px',

        min_height='0px',

        max_width='0px',

        max_height='0px',

        overflow='hidden',

        margin='0px',

        padding='0px'
    )
)


# ================================================================
# 14. 🖥️ OUTPUT MÀN HÌNH
# ================================================================

dm_view_output = widgets.Output(

    layout=widgets.Layout(

        width='100%',

        height='auto',

        min_height='0px',

        max_height='none',

        overflow='visible',

        margin='0px',

        padding='0px',

        flex='0 0 auto'
    )
)

dm_view_output.add_class(
    'app-view-output'
)


# ================================================================
# 15. 🔧 BIẾN WIDGET
# ================================================================

dm_search_box = None

dm_output_message = None

dm_output_form = None

dm_output_table = None


# ================================================================
# 🧹 HÀM XÓA THÔNG BÁO
# ================================================================

def dm_clear_message():

    global dm_output_message

    if dm_output_message is None:
        return

    try:

        # --------------------------------------------------------
        # Cách 1: Xóa toàn bộ output đang lưu trong widget
        # --------------------------------------------------------

        dm_output_message.outputs = ()

    except Exception:

        pass

    try:

        # --------------------------------------------------------
        # Cách 2: Xóa output đang hiển thị
        # --------------------------------------------------------

        dm_output_message.clear_output(
            wait=True
        )

    except Exception:

        pass


# ================================================================
# 16. 📊 HIỂN THỊ BẢNG
# ================================================================

def dm_show_management_table(

    key,

    df=None,

    search_mode=False

):

    if df is None:

        df = dm_TABLE_DATA[key]


    display_df = dm_preserve_table_columns(

        key,

        df.copy()

    )


    display_df = dm_reset_stt(

        display_df

    )


    config = dm_TABLE_CONFIG[key]

    color = dm_TABLE_COLORS[key]


    if display_df.empty:

        with dm_output_table:

            clear_output(

                wait=True

            )

            display(

                HTML("""

                <div style="
                    width:100%;
                    text-align:center;
                    padding:30px;
                    font-size:18px;
                    font-weight:bold;
                    color:#000000;
                    background:#FFE5D8;
                ">

                    🔍 Không tìm thấy dữ liệu

                </div>

                """)

            )

        return


    table_id = (

        f"management_table_{key}"

    )


    # ============================================================
    # 📊 STYLER
    # ============================================================

    styler = (

        display_df

        .style

        .hide(axis='index')

        .set_caption(

            config['caption']

        )

        .set_properties(

            **{

                'text-align':
                    'center !important',

                'vertical-align':
                    'middle !important',

                'padding':
                    '8px !important',

                'color':
                    'white !important',

                'background-color':
                    '#000000 !important',

                'border':
                    '2px solid #666666 !important',

                'font-size':
                    '14px !important'

            }

        )

        .set_table_styles(

            [

                {
                    'selector':
                        'table',

                    'props':
                        [

                            (
                                'border-collapse',
                                'separate !important'
                            ),

                            (
                                'border-spacing',
                                '0 !important'
                            ),

                            (
                                'border-radius',
                                '10px !important'
                            ),

                            (
                                'overflow',
                                'hidden !important'
                            ),

                            (
                                'margin',
                                '0 auto !important'
                            ),

                            (
                                'background-color',
                                '#000000 !important'
                            )

                        ]

                },

                {
                    'selector':
                        'caption',

                    'props':
                        [

                            (
                                'caption-side',
                                'top !important'
                            ),

                            (
                                'font-size',
                                '19px !important'
                            ),

                            (
                                'font-weight',
                                'bold !important'
                            ),

                            (
                                'color',
                                f'{color} !important'
                            ),

                            (
                                'padding',
                                '10px !important'
                            ),

                            (
                                'background-color',
                                '#FFE5D8 !important'
                            ),

                            (
                                'text-align',
                                'center !important'
                            ),

                            (
                                'border',
                                'none !important'
                            )

                        ]

                },

                {
                    'selector':
                        'thead',

                    'props':
                        [

                            (
                                'background-color',
                                f'{color} !important'
                            )

                        ]

                },

                {
                    'selector':
                        'thead th',

                    'props':
                        [

                            (
                                'background-color',
                                f'{color} !important'
                            ),

                            (
                                'color',
                                'white !important'
                            ),

                            (
                                'font-size',
                                '18px !important'
                            ),

                            (
                                'font-weight',
                                'bold !important'
                            ),

                            (
                                'text-align',
                                'center !important'
                            ),

                            (
                                'vertical-align',
                                'middle !important'
                            ),

                            (
                                'padding',
                                '10px !important'
                            ),

                            (
                                'border',
                                '2px solid #666666 !important'
                            ),

                            (
                                'position',
                                'sticky !important'
                            ),

                            (
                                'top',
                                '0px !important'
                            ),

                            (
                                'z-index',
                                '100 !important'
                            )

                        ]

                },

                {
                    'selector':
                        'tbody',

                    'props':
                        [

                            (
                                'background-color',
                                '#000000 !important'
                            )

                        ]

                },

                {
                    'selector':
                        'tbody td',

                    'props':
                        [

                            (
                                'background-color',
                                '#000000 !important'
                            ),

                            (
                                'color',
                                'white !important'
                            ),

                            (
                                'text-align',
                                'center !important'
                            ),

                            (
                                'vertical-align',
                                'middle !important'
                            ),

                            (
                                'border',
                                '2px solid #666666 !important'
                            ),

                            (
                                'padding',
                                '8px !important'
                            ),

                            (
                                'font-size',
                                '14px !important'
                            )

                        ]

                },

                {
                    'selector':
                        'tbody tr:hover td',

                    'props':
                        [

                            (
                                'background-color',
                                f'{color} !important'
                            ),

                            (
                                'color',
                                'white !important'
                            ),

                            (
                                'font-weight',
                                'bold !important'
                            ),

                            (
                                'cursor',
                                'pointer !important'
                            )

                        ]

                },

                {
                    'selector':
                        'tbody tr.dm-selected-row td',

                    'props':
                        [

                            (
                                'background-color',
                                '#FFD700 !important'
                            ),

                            (
                                'color',
                                '#000000 !important'
                            ),

                            (
                                'font-weight',
                                'bold !important'
                            )

                        ]

                },

                {
                    'selector':
                        'tbody tr.dm-selected-row:hover td',

                    'props':
                        [

                            (
                                'background-color',
                                '#FFE866 !important'
                            ),

                            (
                                'color',
                                '#000000 !important'
                            ),

                            (
                                'font-weight',
                                'bold !important'
                            ),

                            (
                                'cursor',
                                'pointer !important'
                            )

                        ]

                },

                {
                    'selector':
                        'tbody td *',

                    'props':
                        [

                            (
                                'color',
                                'white !important'
                            )

                        ]

                }

            ]

        )

    )


    # ============================================================
    # 📏 ĐỘ RỘNG CỘT
    # ============================================================

    for column, width in config.get(

        'column_widths',

        {}

    ).items():

        if column in display_df.columns:

            styler = styler.set_properties(

                subset=[column],

                **{

                    'width': width,

                    'min-width': width,

                    'max-width': width

                }

            )


            if column in [

                'Tên công ty',

                'Chủ hàng',

                'Sà lan'

            ]:

                styler = styler.set_properties(

                    subset=[column],

                    **{

                        'text-align':
                            'left !important'

                    }

                )


    html = styler.to_html()


    # ============================================================
    # 🆔 GÁN ID TABLE
    # ============================================================

    html = re.sub(

        r'<table([^>]*)>',

        f'<table id="{table_id}" '
        f'data-table-key="{key}"\\1>',

        html,

        count=1

    )


    # ============================================================
    # 🎨 CSS TABLE
    # ============================================================

    table_css = f"""

    <style>

    #{table_id} caption {{

        caption-side: top !important;

        color: {color} !important;

        background-color: #FFE5D8 !important;

        font-size: 19px !important;

        font-weight: bold !important;

        text-align: center !important;

        padding: 10px !important;

        border: none !important;

        margin: 0 !important;

    }}


    #{table_id} thead th {{

        background-color:
            {color} !important;

        color:
            white !important;

        font-size:
            18px !important;

        font-weight:
            bold !important;

        text-align:
            center !important;

        vertical-align:
            middle !important;

        border:
            2px solid #666666 !important;

        position:
            sticky !important;

        top:
            0 !important;

        z-index:
            100 !important;

    }}


    #{table_id} tbody tr {{

        background-color:
            #000000 !important;

    }}


    #{table_id} tbody td {{

        background-color:
            #000000 !important;

        color:
            white !important;

        border:
            2px solid #666666 !important;

        text-align:
            center !important;

        vertical-align:
            middle !important;

        white-space:
            normal !important;

        overflow-wrap:
            anywhere !important;

        word-break:
            break-word !important;

    }}


    #{table_id} tbody tr:hover td {{

        background-color:
            {color} !important;

        color:
            white !important;

        font-weight:
            bold !important;

        cursor:
            pointer !important;

    }}


    #{table_id} tbody tr.dm-selected-row td {{

        background-color:
            #FFD700 !important;

        color:
            #000000 !important;

        font-weight:
            bold !important;

    }}


    #{table_id} tbody tr.dm-selected-row:hover td {{

        background-color:
            #FFE866 !important;

        color:
            #000000 !important;

        font-weight:
            bold !important;

        cursor:
            pointer !important;

    }}


    #{table_id} tbody td * {{

        color:
            inherit !important;

        white-space:
            normal !important;

        overflow-wrap:
            anywhere !important;

        word-break:
            break-word !important;

    }}

    </style>

    """


    # ============================================================
    # 🖱️ JAVASCRIPT CHỌN DÒNG
    # ============================================================

    js_template = """

    <script>

    setTimeout(function() {

        const table = document.getElementById(
            "__TABLE_ID__"
        );

        if (!table) return;


        const rows = table.querySelectorAll(
            "tbody tr"
        );


        rows.forEach(function(row, index) {

            row.style.cursor = "pointer";


            row.addEventListener(
                "click",
                function() {


                    rows.forEach(function(r) {

                        r.classList.remove(
                            "dm-selected-row"
                        );

                    });


                    row.classList.add(
                        "dm-selected-row"
                    );


                    google.colab.kernel.invokeFunction(

                        "management_table.select_row",

                        ["__KEY__", index],

                        {}

                    );

                }

            );

        });

    }, 500);

    </script>

    """


    js = (

        js_template

        .replace(

            "__TABLE_ID__",

            table_id

        )

        .replace(

            "__KEY__",

            key

        )

    )


    # ============================================================
    # 📤 HIỂN THỊ
    # ============================================================

    with dm_output_table:

        clear_output(

            wait=True

        )

        display(

            HTML(

                table_css +

                html +

                js

            )

        )


# ================================================================
# 17. 🖱️ CHỌN DÒNG
# ================================================================

def dm_select_row(

    key,

    row_position

):

    try:

        df = dm_SEARCH_RESULTS.get(key)


        if df is None:

            df = dm_TABLE_DATA.get(key)


        if df is None:

            dm_clear_message(key)

            out = dm_MESSAGE_OUTPUTS.get(key)

            if out is not None:

                with out:

                    print(
                        "⚠️ Không có dữ liệu để chọn dòng."
                    )

            return


        if not isinstance(df, pd.DataFrame):

            dm_clear_message(key)

            out = dm_MESSAGE_OUTPUTS.get(key)

            if out is not None:

                with out:

                    print(
                        "⚠️ Dữ liệu bảng không hợp lệ."
                    )

            return


        if (

            row_position is None

            or row_position < 0

            or row_position >= len(df)

        ):

            dm_clear_message(key)

            out = dm_MESSAGE_OUTPUTS.get(key)

            if out is not None:

                with out:

                    print(
                        "⚠️ Dòng được chọn không hợp lệ."
                    )

            return


        original_index = (

            df.index[row_position]

        )


        dm_SELECTED_ROWS[key] = (

            original_index

        )


        # --------------------------------------------------------
        # XÓA THÔNG BÁO CŨ CỦA ĐÚNG DANH MỤC
        # --------------------------------------------------------

        dm_clear_message(key)


        out = dm_MESSAGE_OUTPUTS.get(key)

        if out is not None:

            with out:

                print(

                    f"✅ Đã chọn dòng "
                    f"{row_position + 1}"

                )


    except Exception as e:

        dm_clear_message(key)

        out = dm_MESSAGE_OUTPUTS.get(key)

        if out is not None:

            with out:

                print(

                    f"❌ Lỗi chọn dòng: {e}"

                )


# ================================================================
# 18. 🔗 CALLBACK
# ================================================================

try:

    colab_output.register_callback(

        'management_table.select_row',

        dm_select_row

    )

except Exception:

    pass


# ================================================================
# 19. 🔎 TÌM KIẾM
# ================================================================

def dm_search_data(key):

    keyword_original = (

        dm_search_box.value.strip()

    )


    # ============================================================
    # KHÔNG NHẬP TỪ KHÓA → HIỆN TOÀN BỘ
    # ============================================================

    if not keyword_original:

        dm_SEARCH_RESULTS[key] = (

            dm_preserve_table_columns(

                key,

                dm_TABLE_DATA[key].copy()

            )

        )


        dm_SELECTED_ROWS[key] = None


        dm_clear_message()


        with dm_output_message:

            print(

                f"📋 Hiển thị toàn bộ "

                f"{len(dm_TABLE_DATA[key])} dòng"

            )


        dm_show_management_table(

            key,

            dm_SEARCH_RESULTS[key]

        )

        return


    keyword = (

        keyword_original.lower()

    )


    index_df = dm_SEARCH_INDEX[key]


    mask = index_df.apply(

        lambda col:

            col.str.contains(

                keyword,

                case=False,

                na=False,

                regex=False

            )

    ).any(axis=1)


    dm_SEARCH_RESULTS[key] = (

        dm_TABLE_DATA[key]

        .loc[mask]

        .copy()

    )


    dm_SELECTED_ROWS[key] = None


    dm_clear_message()


    with dm_output_message:

        print(

            f"🔎 Tìm thấy "

            f"{len(dm_SEARCH_RESULTS[key])} "

            f"dòng phù hợp với: "

            f"'{keyword_original}'"

        )


    dm_show_management_table(

        key,

        dm_SEARCH_RESULTS[key]

    )


# ================================================================
# 20. 👁️ HIỆN TẤT CẢ
# ================================================================

def dm_show_all_data(key):

    # ============================================================
    # 1. RESET KẾT QUẢ TÌM KIẾM
    # ============================================================

    dm_SEARCH_RESULTS[key] = (

        dm_preserve_table_columns(

            key,

            dm_TABLE_DATA[key].copy()

        )

    )


    # ============================================================
    # 2. BỎ DÒNG ĐANG CHỌN
    # ============================================================

    dm_SELECTED_ROWS[key] = None


    # ============================================================
    # 3. XÓA Ô TÌM KIẾM
    # ============================================================

    search_box = dm_SEARCH_BOXES.get(key)

    if search_box is not None:

        try:

            search_box.unobserve_all()

        except Exception:

            pass

        search_box.value = ''

        # Không cần giữ observer cũ ở đây


    # ============================================================
    # 4. XÓA FORM
    # ============================================================

    dm_clear_form()


    # ============================================================
    # 5. 🔥 XÓA CHÍNH XÁC THÔNG BÁO CỦA DANH MỤC
    # ============================================================

    message_out = dm_MESSAGE_OUTPUTS.get(key)

    if message_out is not None:

        try:

            message_out.clear_output(
                wait=False
            )

        except Exception:
            pass

        try:

            message_out.outputs = ()

        except Exception:
            pass


    # ============================================================
    # 6. XÓA GLOBAL OUTPUT NẾU ĐANG TRỎ ĐÚNG DANH MỤC
    # ============================================================

    global dm_output_message

    if dm_output_message is message_out:

        try:

            dm_output_message.clear_output(
                wait=False
            )

        except Exception:
            pass

        try:

            dm_output_message.outputs = ()

        except Exception:
            pass


    # ============================================================
    # 7. HIỂN THỊ LẠI TOÀN BỘ BẢNG
    # ============================================================

    dm_show_management_table(

        key,

        dm_SEARCH_RESULTS[key],

        search_mode=False

    )


# ================================================================
# 21. 💾 GHI DATAFRAME VÀO EXCEL
# ================================================================

def dm_write_dataframe_to_excel(

    file_path,

    df

):

    wb = openpyxl.load_workbook(
        file_path
    )

    ws = wb.active


    old_last_row = ws.max_row

    old_max_col = ws.max_column


    max_col = max(

        old_max_col,

        len(df.columns)
    )


    source_row_styles = {}


    if old_last_row >= 2:

        for col_idx in range(

            1,

            max_col + 1
        ):

            source_cell = ws.cell(

                row=2,

                column=col_idx
            )


            source_row_styles[col_idx] = {

                'font':
                    copy(source_cell.font),

                'fill':
                    copy(source_cell.fill),

                'border':
                    copy(source_cell.border),

                'alignment':
                    copy(source_cell.alignment),

                'number_format':
                    source_cell.number_format,

                'protection':
                    copy(source_cell.protection)
            }


    if old_last_row >= 2:

        for row in ws.iter_rows(

            min_row=2,

            max_row=old_last_row,

            min_col=1,

            max_col=max_col
        ):

            for cell in row:

                cell.value = None


    for row_idx, row in enumerate(

        df.itertuples(

            index=False,

            name=None
        ),

        start=2
    ):

        for col_idx, value in enumerate(

            row,

            start=1
        ):

            cell = ws.cell(

                row=row_idx,

                column=col_idx
            )


            if (

                row_idx > old_last_row

                and col_idx in source_row_styles
            ):

                style = source_row_styles[col_idx]


                cell.font = copy(
                    style['font']
                )

                cell.fill = copy(
                    style['fill']
                )

                cell.border = copy(
                    style['border']
                )

                cell.alignment = copy(
                    style['alignment']
                )

                cell.number_format = (
                    style['number_format']
                )

                cell.protection = copy(
                    style['protection']
                )


            if pd.isna(value):

                cell.value = None

            else:

                cell.value = value


    new_last_row = len(df) + 1


    if old_last_row > new_last_row:

        for row_idx in range(

            new_last_row + 1,

            old_last_row + 1
        ):

            for col_idx in range(

                1,

                max_col + 1
            ):

                ws.cell(

                    row=row_idx,

                    column=col_idx
                ).value = None


    wb.save(
        file_path
    )


# ================================================================
# 22. 💾 LƯU DATAFRAME
# ================================================================

def dm_save_dataframe_to_excel(key):

    file_path = dm_TABLE_FILES[key]


    try:

        df = dm_preserve_table_columns(

            key,

            dm_TABLE_DATA[key].copy()
        )


        df = dm_reset_stt(
            df
        )


        dm_write_dataframe_to_excel(

            file_path,

            df
        )


        dm_TABLE_DATA[key] = df.copy()


        dm_SEARCH_RESULTS[key] = (
            df.copy()
        )


        dm_SELECTED_ROWS[key] = None


        dm_refresh_search_index(key)


        return True


    except Exception as e:

        dm_output_message.clear_output(
            wait=True
        )

        with dm_output_message:

            print(
                f"❌ Lỗi lưu Excel: {e}"
            )

        return False


# ================================================================
# 23. 🔄 REFRESH DATA
# ================================================================

def dm_refresh_table_data(key):

    dm_TABLE_DATA[key] = (

        dm_preserve_table_columns(

            key,

            dm_reset_stt(

                dm_TABLE_DATA[key].copy()
            )
        )
    )


    dm_SEARCH_RESULTS[key] = (
        dm_TABLE_DATA[key].copy()
    )


    dm_SELECTED_ROWS[key] = None


    dm_refresh_search_index(key)


# ================================================================
# 24. 🧹 XÓA FORM
# ================================================================

def dm_clear_form():

    if dm_output_form is not None:

        dm_output_form.children = ()


# ================================================================
# 25. ➕ THÊM DỮ LIỆU
# ================================================================

def dm_add_data(key):

    dm_clear_form()


    dm_output_message.clear_output(
        wait=True
    )


    dm_TABLE_DATA[key] = (

        dm_preserve_table_columns(

            key,

            dm_TABLE_DATA[key].copy()
        )
    )


    df = dm_TABLE_DATA[key]

    inputs = {}

    form_children = []


    for column in df.columns:

        if column == 'STT':

            continue


        widget = widgets.Text(

            description=f'{column}:',

            layout=widgets.Layout(

                width='650px',

                height='38px',

                min_height='38px',

                max_height='38px',

                margin='2px auto',

                flex='0 0 38px'
            ),

            style={

                'description_width':
                    '150px'
            }
        )


        inputs[column] = widget

        form_children.append(
            widget
        )


    btn_confirm = widgets.Button(

        description='💾 Xác nhận thêm',

        layout=widgets.Layout(

            width='auto',

            min_width='0px',

            height='35px',

            min_height='35px',

            max_height='35px',

            margin='0px',

            flex='0 0 auto'
        )
    )


    btn_cancel = widgets.Button(

        description='❌ Hủy',

        layout=widgets.Layout(

            width='auto',

            min_width='0px',

            height='35px',

            min_height='35px',

            max_height='35px',

            margin='0px',

            flex='0 0 auto'
        )
    )


    button_box = widgets.HBox(

        [

            btn_confirm,

            btn_cancel

        ],

        layout=widgets.Layout(

            width='100%',

            height='45px',

            min_height='45px',

            max_height='45px',

            justify_content='center',

            align_items='center',

            gap='10px',

            margin='6px 0 3px 0',

            flex='0 0 45px',

            overflow='visible'
        )
    )


    button_box.add_class(
        'form-button-box'
    )


    form_children.append(
        button_box
    )


    dm_output_form.children = tuple(
        form_children
    )


    def confirm_add(_):

        new_data = {}


        for column in df.columns:

            if column == 'STT':

                continue


            new_data[column] = (

                inputs[column]
                .value
                .strip()
            )


        duplicate_column = None


        for col in [

            'Tên công ty',

            'Chủ hàng',

            'Sà lan'
        ]:

            if col in df.columns:

                duplicate_column = col

                break


        if duplicate_column:

            value = new_data.get(

                duplicate_column,

                ''
            )


            if value:

                exists = (

                    df[
                        duplicate_column
                    ]

                    .fillna('')

                    .astype(str)

                    .str.strip()

                    .str.lower()

                    .eq(
                        value.lower()
                    )

                    .any()
                )


                if exists:

                    dm_output_message.clear_output(
                        wait=True
                    )

                    with dm_output_message:

                        print(

                            f"⚠️ "
                            f"{duplicate_column} "
                            f"'{value}' "
                            f"đã tồn tại."
                        )

                    return


        new_row_data = {}


        for column in dm_TABLE_COLUMNS[key]:

            if column == 'STT':

                new_row_data[column] = pd.NA

            else:

                new_row_data[column] = (

                    new_data.get(

                        column,

                        pd.NA
                    )
                )


        new_row = pd.DataFrame(

            [new_row_data],

            columns=dm_TABLE_COLUMNS[key]
        )


        base_df = (

            dm_preserve_table_columns(

                key,

                dm_TABLE_DATA[key].copy()
            )
        )


        dm_TABLE_DATA[key] = pd.concat(

            [

                base_df,

                new_row

            ],

            ignore_index=True
        )


        dm_TABLE_DATA[key] = (

            dm_preserve_table_columns(

                key,

                dm_reset_stt(

                    dm_TABLE_DATA[key]
                )
            )
        )


        if dm_save_dataframe_to_excel(key):

            dm_output_message.clear_output(
                wait=True
            )

            with dm_output_message:

                print(

                    "✅ Đã thêm dữ liệu "
                    "và lưu vào Excel."
                )


        dm_clear_form()


        dm_show_management_table(

            key,

            dm_TABLE_DATA[key]
        )


    def cancel_add(_):

        dm_clear_form()

        dm_output_message.clear_output(
            wait=True
        )


    btn_confirm.on_click(
        confirm_add
    )

    btn_cancel.on_click(
        cancel_add
    )


# ================================================================
# 26. ✏️ SỬA DỮ LIỆU
# ================================================================

def dm_edit_data(key):

    selected = dm_SELECTED_ROWS[key]


    if selected is None:

        dm_output_message.clear_output(
            wait=True
        )

        with dm_output_message:

            print(

                "⚠️ Vui lòng click chọn "
                "một dòng trước khi sửa."
            )

        return


    dm_TABLE_DATA[key] = (

        dm_preserve_table_columns(

            key,

            dm_TABLE_DATA[key].copy()
        )
    )


    df = dm_TABLE_DATA[key]


    if selected not in df.index:

        dm_output_message.clear_output(
            wait=True
        )

        with dm_output_message:

            print(

                "❌ Không tìm thấy dòng đã chọn."
            )

        return


    current_row = df.loc[selected]


    dm_clear_form()


    dm_output_message.clear_output(
        wait=True
    )


    inputs = {}

    form_children = []


    for column in df.columns:

        value = current_row[column]


        if pd.isna(value):

            value = ''


        widget = widgets.Text(

            value=str(value),

            description=f'{column}:',

            layout=widgets.Layout(

                width='650px',

                height='38px',

                min_height='38px',

                max_height='38px',

                margin='2px auto',

                flex='0 0 38px'
            ),

            style={

                'description_width':
                    '150px'
            }
        )


        if column == 'STT':

            widget.disabled = True


        inputs[column] = widget

        form_children.append(
            widget
        )


    btn_confirm = widgets.Button(

        description='💾 Xác nhận sửa',

        layout=widgets.Layout(

            width='auto',

            min_width='0px',

            height='35px',

            min_height='35px',

            max_height='35px',

            margin='0px',

            flex='0 0 auto'
        )
    )


    btn_cancel = widgets.Button(

        description='❌ Hủy',

        layout=widgets.Layout(

            width='auto',

            min_width='0px',

            height='35px',

            min_height='35px',

            max_height='35px',

            margin='0px',

            flex='0 0 auto'
        )
    )


    button_box = widgets.HBox(

        [

            btn_confirm,

            btn_cancel

        ],

        layout=widgets.Layout(

            width='100%',

            height='45px',

            min_height='45px',

            max_height='45px',

            justify_content='center',

            align_items='center',

            gap='10px',

            margin='6px 0 3px 0',

            flex='0 0 45px',

            overflow='visible'
        )
    )


    button_box.add_class(
        'form-button-box'
    )


    form_children.append(
        button_box
    )


    dm_output_form.children = tuple(
        form_children
    )


    def confirm_edit(_):

        for column in dm_TABLE_COLUMNS[key]:

            if column == 'STT':

                continue


            dm_TABLE_DATA[key].at[

                selected,

                column

            ] = (

                inputs[column]
                .value
                .strip()
            )


        dm_TABLE_DATA[key] = (

            dm_preserve_table_columns(

                key,

                dm_reset_stt(

                    dm_TABLE_DATA[key]
                )
            )
        )


        if dm_save_dataframe_to_excel(key):

            dm_output_message.clear_output(
                wait=True
            )

            with dm_output_message:

                print(

                    "✅ Đã sửa dữ liệu "
                    "và lưu vào Excel."
                )


        dm_SELECTED_ROWS[key] = None


        dm_clear_form()


        dm_show_management_table(

            key,

            dm_TABLE_DATA[key]
        )


    def cancel_edit(_):

        dm_clear_form()

        dm_output_message.clear_output(
            wait=True
        )


    btn_confirm.on_click(
        confirm_edit
    )

    btn_cancel.on_click(
        cancel_edit
    )


# ================================================================
# 27. 🗑️ XÓA DỮ LIỆU
# ================================================================

def dm_delete_data(key):

    selected = dm_SELECTED_ROWS[key]


    if selected is None:

        dm_output_message.clear_output(
            wait=True
        )

        with dm_output_message:

            print(

                "⚠️ Vui lòng click chọn "
                "một dòng trước khi xóa."
            )

        return


    dm_TABLE_DATA[key] = (

        dm_preserve_table_columns(

            key,

            dm_TABLE_DATA[key].copy()
        )
    )


    df = dm_TABLE_DATA[key]


    if selected not in df.index:

        dm_output_message.clear_output(
            wait=True
        )

        with dm_output_message:

            print(

                "❌ Không tìm thấy dòng đã chọn."
            )

        return


    display_name = ''


    for col in [

        'Tên công ty',

        'Chủ hàng',

        'Sà lan'
    ]:

        if col in df.columns:

            value = df.loc[

                selected,

                col
            ]


            if not pd.isna(value):

                display_name = str(value)

                break


    dm_clear_form()


    dm_output_message.clear_output(
        wait=True
    )


    with dm_output_message:

        display(

            HTML(
                f"""
                <div style="
                    text-align:center;
                    color:#000000;
                    font-size:15px;
                    font-weight:600;
                    padding:3px 0;
                    margin:0;
                    line-height:1.2;
                ">
                    ⚠️ Bạn có chắc muốn xóa
                    '<b>{display_name}</b>' không?
                </div>
                """
            )
        )


        btn_yes = widgets.Button(

            description='🗑️ Xóa',

            layout=widgets.Layout(

                width='auto',

                min_width='0px',

                height='35px',

                min_height='35px',

                max_height='35px',

                flex='0 0 auto'
            )
        )


        btn_no = widgets.Button(

            description='❌ Hủy',

            layout=widgets.Layout(

                width='auto',

                min_width='0px',

                height='35px',

                min_height='35px',

                max_height='35px',

                flex='0 0 auto'
            )
        )


        confirm_box = widgets.HBox(

            [

                btn_yes,

                btn_no

            ],

            layout=widgets.Layout(

                width='100%',

                height='45px',

                min_height='45px',

                justify_content='center',

                align_items='center',

                gap='10px',

                margin='3px 0',

                flex='0 0 45px',

                overflow='visible'
            )
        )


        display(
            confirm_box
        )


        def confirm_delete(_):

            dm_TABLE_DATA[key] = (

                dm_TABLE_DATA[key]

                .drop(
                    index=selected
                )

                .reset_index(
                    drop=True
                )
            )


            dm_TABLE_DATA[key] = (

                dm_preserve_table_columns(

                    key,

                    dm_reset_stt(

                        dm_TABLE_DATA[key]
                    )
                )
            )


            if dm_save_dataframe_to_excel(key):

                dm_output_message.clear_output(
                    wait=True
                )

                with dm_output_message:

                    print(

                        "✅ Đã xóa dữ liệu "
                        "và lưu vào Excel."
                    )


            dm_SELECTED_ROWS[key] = None


            dm_clear_form()


            dm_show_management_table(

                key,

                dm_TABLE_DATA[key]
            )


        def cancel_delete(_):

            dm_output_message.clear_output(
                wait=True
            )


        btn_yes.on_click(
            confirm_delete
        )

        btn_no.on_click(
            cancel_delete
        )


# ================================================================
# 28. 📤 KẾT XUẤT KẾT QUẢ TÌM KIẾM
# ================================================================

def dm_export_search_result(key):

    keyword = dm_search_box.value.strip()


    if not keyword:

        dm_output_message.clear_output(
            wait=True
        )

        with dm_output_message:

            print(

                "⚠️ Vui lòng nhập từ khóa "
                "trước khi kết xuất."
            )

        return


    result = (

        dm_preserve_table_columns(

            key,

            dm_SEARCH_RESULTS[key].copy()
        )
    )


    result = dm_reset_stt(
        result
    )


    if result.empty:

        dm_output_message.clear_output(
            wait=True
        )

        with dm_output_message:

            print(

                "⚠️ Không có dữ liệu để kết xuất."
            )

        return


    try:

        source_file = (
            dm_TABLE_FILES[key]
        )


        output_file = (

            f'/content/'
            f'Ket_qua_tim_kiem_{key}.xlsx'
        )


        shutil.copy2(

            source_file,

            output_file
        )


        wb = openpyxl.load_workbook(
            output_file
        )


        ws = wb.active


        old_last_row = ws.max_row

        old_max_col = ws.max_column


        source_styles = {}


        if old_last_row >= 2:

            for col_idx in range(

                1,

                old_max_col + 1
            ):

                source = ws.cell(

                    row=2,

                    column=col_idx
                )


                source_styles[col_idx] = {

                    'font':
                        copy(source.font),

                    'fill':
                        copy(source.fill),

                    'border':
                        copy(source.border),

                    'alignment':
                        copy(source.alignment),

                    'number_format':
                        source.number_format,

                    'protection':
                        copy(source.protection)
                }


        if old_last_row >= 2:

            for row in ws.iter_rows(

                min_row=2,

                max_row=old_last_row,

                min_col=1,

                max_col=old_max_col
            ):

                for cell in row:

                    cell.value = None


        for new_row_idx, (_, row) in enumerate(

            result.iterrows(),

            start=2
        ):

            for col_idx, value in enumerate(

                row.tolist(),

                start=1
            ):

                cell = ws.cell(

                    row=new_row_idx,

                    column=col_idx
                )


                if pd.isna(value):

                    cell.value = None

                else:

                    cell.value = value


                if col_idx in source_styles:

                    style = source_styles[col_idx]


                    cell.font = copy(
                        style['font']
                    )

                    cell.fill = copy(
                        style['fill']
                    )

                    cell.border = copy(
                        style['border']
                    )

                    cell.alignment = copy(
                        style['alignment']
                    )

                    cell.number_format = (
                        style['number_format']
                    )

                    cell.protection = copy(
                        style['protection']
                    )


        wb.save(
            output_file
        )


        dm_output_message.clear_output(
            wait=True
        )

        with dm_output_message:

            print(

                f"✅ Đã kết xuất "
                f"{len(result)} dòng."
            )


        files.download(
            output_file
        )


    except Exception as e:

        dm_output_message.clear_output(
            wait=True
        )

        with dm_output_message:

            print(
                f"❌ Lỗi kết xuất: {e}"
            )


# ================================================================
# 29. 🔙 NAVIGATION MÀN HÌNH BẢNG
# ================================================================

def dm_create_table_navigation():

    back_btn = widgets.Button(

        description='⬅️ BACK',

        layout=widgets.Layout(

            width='auto',

            min_width='0px',

            height='38px',

            min_height='38px',

            max_height='38px',

            flex='0 0 auto'
        )
    )


    home_btn = widgets.Button(

        description='🏠 MENU CHÍNH',

        layout=widgets.Layout(

            width='auto',

            min_width='0px',

            height='38px',

            min_height='38px',

            max_height='38px',

            flex='0 0 auto'
        )
    )


    back_btn.add_class(
        'app-nav-button'
    )

    home_btn.add_class(
        'app-nav-button'
    )


    back_btn.on_click(

        lambda _:
            dm_show_category_menu()
    )


    home_btn.on_click(

        lambda _:
            app_show_home()
    )


    navigation = widgets.HBox(

        [

            back_btn,

            home_btn

        ],

        layout=widgets.Layout(

            width='100%',

            justify_content='flex-start',

            align_items='center',

            gap='8px',

            margin='0',

            padding='0px',

            height='38px',

            min_height='38px',

            overflow='visible',

            flex='0 0 38px'
        )
    )


    navigation.add_class(
        'dm-table-navigation'
    )


    return navigation


# ================================================================
# 30. 🖥️ TẠO GIAO DIỆN QUẢN LÝ
# ================================================================

def dm_create_management_ui(key):

    global dm_search_box
    global dm_output_message
    global dm_output_form
    global dm_output_table


    # ============================================================
    # Ô TÌM KIẾM
    # ============================================================

    dm_search_box = widgets.Text(

        placeholder=
            '⌨ Nhập dữ liệu cần tìm...',

        description='',

        layout=widgets.Layout(

            width='180px',

            min_width='180px',

            height='38px',

            min_height='38px',

            max_height='38px',

            flex='0 0 180px'
        )
    )


    # ============================================================
    # BUTTON TÌM KIẾM
    # ⭐ AUTO WIDTH
    # ============================================================

    btn_search = widgets.Button(

        description='🔎 Tìm kiếm',

        layout=widgets.Layout(

            width='auto',

            min_width='0px',

            height='38px',

            min_height='38px',

            max_height='38px',

            flex='0 0 auto'
        )
    )


    # ============================================================
    # BUTTON HIỆN TẤT CẢ
    # ⭐ AUTO WIDTH
    # ============================================================

    btn_show_all = widgets.Button(

        description='🔄 Hiện tất cả',

        layout=widgets.Layout(

            width='auto',

            min_width='0px',

            height='38px',

            min_height='38px',

            max_height='38px',

            flex='0 0 auto'
        )
    )


    # ============================================================
    # BUTTON THÊM
    # ⭐ AUTO WIDTH
    # ============================================================

    btn_add = widgets.Button(

        description='➕ Thêm',

        layout=widgets.Layout(

            width='auto',

            min_width='0px',

            height='38px',

            min_height='38px',

            max_height='38px',

            flex='0 0 auto'
        )
    )


    # ============================================================
    # BUTTON SỬA
    # ⭐ AUTO WIDTH
    # ============================================================

    btn_edit = widgets.Button(

        description='✏️ Sửa',

        layout=widgets.Layout(

            width='auto',

            min_width='0px',

            height='38px',

            min_height='38px',

            max_height='38px',

            flex='0 0 auto'
        )
    )


    # ============================================================
    # BUTTON XÓA
    # ⭐ AUTO WIDTH
    # ============================================================

    btn_delete = widgets.Button(

        description='🗑️ Xóa',

        layout=widgets.Layout(

            width='auto',

            min_width='0px',

            height='38px',

            min_height='38px',

            max_height='38px',

            flex='0 0 auto'
        )
    )


    # ============================================================
    # BUTTON KẾT XUẤT
    # ⭐ AUTO WIDTH
    # ============================================================

    btn_export = widgets.Button(

        description='📤 Kết xuất',

        layout=widgets.Layout(

            width='auto',

            min_width='0px',

            height='38px',

            min_height='38px',

            max_height='38px',

            flex='0 0 auto'
        )
    )


    # ============================================================
    # TOOLBAR
    # ============================================================

    toolbar = widgets.HBox(

        [

            dm_search_box,

            btn_search,

            btn_show_all,

            btn_add,

            btn_edit,

            btn_delete,

            btn_export

        ],

        layout=widgets.Layout(

            width='100%',

            height='auto',

            min_height='42px',

            display='flex',

            flex_flow='row wrap',

            justify_content='center',

            align_items='center',

            gap='8px',

            margin='5px auto 10px auto',

            overflow='visible',

            flex='0 0 auto'
        )
    )


    toolbar.add_class(
        'management-toolbar'
    )


    # ============================================================
    # OUTPUT MESSAGE
    # ============================================================

    dm_output_message = widgets.Output(

        layout=widgets.Layout(

            width='100%',

            height='auto',

            min_height='0px',

            max_height='none',

            overflow='visible',

            margin='0px',

            padding='0px',

            flex='0 0 auto'
        )
    )


    dm_output_message.add_class(
        'management-message'
    )

    dm_MESSAGE_OUTPUTS[key] = dm_output_message


    # ============================================================
    # FORM THÊM / SỬA
    # ============================================================

    dm_output_form = widgets.VBox(

        [],

        layout=widgets.Layout(

            width='100%',

            height='auto',

            min_height='0px',

            max_height='none',

            overflow='visible',

            margin='0px',

            padding='0px',

            flex='0 0 auto',

            align_items='stretch'
        )
    )


    dm_output_form.add_class(
        'management-form'
    )


    # ============================================================
    # OUTPUT TABLE
    # ============================================================

    dm_output_table = widgets.Output(

        layout=widgets.Layout(

            width='100%',

            height='500px',

            min_height='500px',

            max_height='500px',

            overflow='auto',

            margin='0px',

            padding='0px',

            flex='0 0 500px'
        )
    )


    dm_output_table.add_class(
        'management-table-scroll'
    )


    # ============================================================
    # EVENTS
    # ============================================================

    btn_search.on_click(

        lambda _:
            dm_search_data(key)
    )


    btn_show_all.on_click(

        lambda _:
            dm_show_all_data(key)
    )


    btn_add.on_click(

        lambda _:
            dm_add_data(key)
    )


    btn_edit.on_click(

        lambda _:
            dm_edit_data(key)
    )


    btn_delete.on_click(

        lambda _:
            dm_delete_data(key)
    )


    btn_export.on_click(

        lambda _:
            dm_export_search_result(key)
    )


    dm_search_box.on_submit(

        lambda _:
            dm_search_data(key)
    )


    # ============================================================
    # NAVIGATION
    # ============================================================

    navigation = (
        dm_create_table_navigation()
    )


    # ============================================================
    # UI BẢNG
    # ============================================================

    ui = widgets.VBox(

        [

            navigation,

            toolbar,

            dm_output_message,

            dm_output_form,

            dm_output_table

        ],

        layout=widgets.Layout(

            width='100%',

            height='auto',

            min_height='0px',

            max_height='none',

            margin='0 auto',

            padding='0px',

            overflow='visible',

            flex_flow='column',

            align_items='stretch'
        )
    )


    ui.add_class(
        'management-container'
    )


    dm_UI_CACHE[key] = ui


    # ============================================================
    # HIỂN THỊ BẢNG
    # ============================================================

    dm_show_management_table(

        key,

        dm_TABLE_DATA[key]
    )


    return ui


# ================================================================
# 31. 🏢🚢🚤 3 BUTTON DANH MỤC
# ================================================================

dm_main_output = widgets.Output()


# ================================================================
# 🏢 CÔNG TY
# ================================================================

dm_btn_cty = widgets.Button(

    description='🏢 Công ty',

    layout=widgets.Layout(

        width='200px',

        height='60px',

        min_height='60px',

        flex='0 0 200px'
    ),

    style=widgets.ButtonStyle(

        button_color='#E33539',

        font_weight='bold'
    )
)

dm_btn_cty.add_class(
    'main-menu-button'
)


# ================================================================
# 🚢 CHỦ HÀNG
# ================================================================

dm_btn_chuhang = widgets.Button(

    description='🚢 Chủ hàng',

    layout=widgets.Layout(

        width='200px',

        height='60px',

        min_height='60px',

        flex='0 0 200px'
    ),

    style=widgets.ButtonStyle(

        button_color='#EE6457',

        font_weight='bold'
    )
)

dm_btn_chuhang.add_class(
    'main-menu-button'
)


# ================================================================
# 🚤 SÀ LAN
# ================================================================

dm_btn_salan = widgets.Button(

    description='🚤 Sà lan',

    layout=widgets.Layout(

        width='200px',

        height='60px',

        min_height='60px',

        flex='0 0 200px'
    ),

    style=widgets.ButtonStyle(

        button_color='#FF9190',

        font_weight='bold'
    )
)

dm_btn_salan.add_class(
    'main-menu-button'
)


# ================================================================
# 32. 📌 TIÊU ĐỀ DANH MỤC
# ================================================================

dm_category_title = widgets.HTML(

    value="""

    <div class="category-title">

        1️⃣ DANH MỤC

    </div>

    """
)


# ================================================================
# 33. 📐 MENU 3 BUTTON
# ================================================================

dm_main_menu = widgets.HBox(

    [

        dm_btn_cty,

        dm_btn_chuhang,

        dm_btn_salan

    ],

    layout=widgets.Layout(

        width='100%',

        display='flex',

        justify_content='center',

        align_items='center',

        gap='18px',

        margin='0 auto',

        height='auto',

        min_height='60px',

        overflow='visible',

        flex='0 0 auto'
    )
)

dm_main_menu.add_class(
    'main-menu-container'
)


# ================================================================
# 34. 🔙 NAVIGATION DANH MỤC
# ================================================================

dm_category_navigation = widgets.HBox(

    [],

    layout=widgets.Layout(

        width='100%',

        display='flex',

        flex_flow='row',

        align_items='center',

        gap='8px',

        margin='0 0 6px 0',

        padding='0px',

        height='38px',

        min_height='38px',

        overflow='visible',

        flex='0 0 38px'
    )
)

dm_category_navigation.add_class(
    'dm-table-navigation'
)


dm_category_back = widgets.Button(

    description='⬅️ BACK',

    layout=widgets.Layout(

        width='auto',

        min_width='0px',

        height='38px',

        min_height='38px',

        max_height='38px',

        flex='0 0 auto'
    )
)

dm_category_back.add_class(
    'app-nav-button'
)


dm_category_home = widgets.Button(

    description='🏠 MENU CHÍNH',

    layout=widgets.Layout(

        width='auto',

        min_width='0px',

        height='38px',

        min_height='38px',

        max_height='38px',

        flex='0 0 auto'
    )
)

dm_category_home.add_class(
    'app-nav-button'
)


dm_category_navigation.children = (

    dm_category_back,

    dm_category_home
)


dm_category_back.on_click(

    lambda _:
        app_show_home()
)


dm_category_home.on_click(

    lambda _:
        app_show_home()
)


# ================================================================
# 35. 📊 HEADER DANH MỤC
# ================================================================

dm_main_header = widgets.VBox(

    [

        dm_category_navigation,

        dm_category_title,

        dm_main_menu

    ],

    layout=widgets.Layout(

        width='100%',

        height='auto',

        min_height='0px',

        max_height='none',

        margin='0 auto',

        padding='0px',

        overflow='visible',

        flex_flow='column',

        flex='0 0 auto'
    )
)


dm_main_header.add_class(
    'main-header'
)


# ================================================================
# 36. 📊 HIỂN THỊ MENU DANH MỤC
# ================================================================

def dm_show_category_menu():

    with dm_view_output:

        clear_output(
            wait=True
        )

        display(
            dm_main_header
        )


# ================================================================
# 37. 🔄 MỞ BẢNG
# ================================================================

def dm_open_table(key):

    with dm_view_output:

        clear_output(
            wait=True
        )

        display(

            dm_create_management_ui(key)
        )


# ================================================================
# 38. 🔗 BUTTON CÔNG TY
# ================================================================

dm_btn_cty.on_click(

    lambda _:

        dm_open_table(
            'cty_hangtau'
        )
)


# ================================================================
# 39. 🔗 BUTTON CHỦ HÀNG
# ================================================================

dm_btn_chuhang.on_click(

    lambda _:

        dm_open_table(
            'chuhang_hangtau'
        )
)


# ================================================================
# 40. 🔗 BUTTON SÀ LAN
# ================================================================

dm_btn_salan.on_click(

    lambda _:

        dm_open_table(
            'salan'
        )
)


# ================================================================
# 41. 🖥️ GIAO DIỆN DANH MỤC
# ================================================================

dm_main_ui = widgets.VBox(

    [

        dm_css_widget,

        dm_view_output

    ],

    layout=widgets.Layout(

        width='100%',

        height='auto',

        min_height='0px',

        max_height='none',

        margin='0 auto',

        padding='0px',

        overflow='visible',

        flex_flow='column',

        flex='0 0 auto'
    )
)


dm_main_ui.add_class(
    'management-container'
)

```
